```python
# creating dataframe from list of tuples
data_tuples = [
    ("Alice", 25, "New York"),
    ("Bob", 30, "Los Angeles"),
    ("Charlie", 35, "Chicago"),
    ("David", 40, "Houston"),
    ("Eve", 45, "Phoenix"),
    ("Frank", 50, "Philadelphia"),
    ("Grace", 55, "San Antonio"),
    ("Heidi", 60, "San Diego"),
    ("Ivan", 65, "Dallas"),
    ("Judy", 70, "San Jose")
]
df_tuples = pd.DataFrame(data_tuples, columns=["Name", "Age", "City"])
```

```python
# search for a specific name
df_tuples[df_tuples["Name"] == "Alice"]
```

```python
#manipulating dataframes
# adding a new column
df_tuples["Age in 5 Years"] = df_tuples["Age"] + 5
# filtering rows based on a condition
df_filtered = df_tuples[df_tuples["Age"] > 30]
# calculating the mean age
mean_age = df_tuples["Age"].mean()
# grouping by city and calculating the average age
grouped = df_tuples.groupby("City")["Age"].mean()
# sorting by age
sorted_df = df_tuples.sort_values(by="Age", ascending=True)
# saving to a CSV file
df_tuples.to_csv("people.csv", index=False)
# loading from a CSV file
df_loaded = pd.read_csv("people.csv")
# displaying the first few rows of the dataframe
df_loaded.head()
# displaying the last few rows of the dataframe
df_loaded.tail()
# getting summary statistics
summary_stats = df_tuples.describe()
# getting the data types of each column
data_types = df_tuples.dtypes
# checking for missing values
missing_values = df_tuples.isnull().sum()
# renaming columns
df_tuples.rename(columns={"Age in 5 Years": "Age in Five Years"}, inplace=True)
# dropping a column
df_tuples.drop(columns=["Age in Five Years"], inplace=True)
# dropping rows with missing values
df_dropped = df_tuples.dropna()
# filling missing values with a specific value
df_filled = df_tuples.fillna(0)
# applying a function to a column
df_tuples["Age Squared"] = df_tuples["Age"].apply(lambda x: x**2)
# applying a function to multiple columns
df_tuples["Name Length"] = df_tuples["Name"].apply(len)
# applying a function to the entire dataframe
df_tuples["Age Category"] = df_tuples["Age"].apply(lambda x: "Young" if x < 30 else "Middle-aged" if x < 60 else "Senior")
# merging two dataframes
df_merged = pd.merge(df, df_tuples, on=["Name", "Age", "City"], how="inner")
# concatenating two dataframes
df_concatenated = pd.concat([df, df_tuples], ignore_index=True)
# pivoting a dataframe
df_pivoted = df_tuples.pivot(index="Name", columns="City", values="Age")
# melting a dataframe
df_melted = df_tuples.melt(id_vars=["Name"], value_vars=["Age", "City"], var_name="Variable", value_name="Value")
```

```python
# downloading a CSV file from a URL and loading it into a dataframe
url = (
    "https://raw.githubusercontent.com/pandas-dev/pandas/main/pandas/tests/io/data/csv/tips.csv"
)
```

```python
tips_df = pd.read_csv(url)
# tips_df.head()
# tips_df.shape
# tips_df
```

```python
print(tips_df)
```

```python
## Sql Query
# SELECT total_bill, tip, smoker, time FROM tips;

# Equivalent pandas code:
tips_df[["total_bill", "tip", "smoker", "time"]]    

# In SQL, you can add a calculated column:
# SELECT *, tip/total_bill as tip_rate FROM tips;

# Equivalent pandas code:
tips_df["tip_rate"] = tips_df["tip"] / tips_df["total_bill"]

tips_df[["total_bill", "tip", "smoker", "time", "tip_rate"]]

# SELECT * FROM tips WHERE time = 'Dinner';
# equivalent pandas code:
tips_df[tips_df["time"] == "Dinner"]

# SELECT * FROM tips WHERE time = 'Dinner' AND tip > 5.00;
# equivalent pandas code:
tips_df[(tips_df["time"] == "Dinner") & (tips_df["tip"] > 5.00)]

# 
```

```python
# SELECT sex, count(*) FROM tips GROUP BY sex;
# /*
# Female     87
# Male      157
# */
tips_df.groupby("sex").size()
tips_df.groupby("sex").count()
tips_df.groupby("sex")["total_bill"].count()
```

```python
# SELECT day, AVG(tip), COUNT(*) FROM tips GROUP BY day;
# /*
# Fri   2.734737   19
# Sat   2.993103   87
# Sun   3.255132   76
# Thu  2.771452   62
# */
tips_df.groupby("day").agg({"tip": "mean", "day": "size"})
```
