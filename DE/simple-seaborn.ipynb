Descriptive Statistics

Descriptive statistics summarize and describe the main features of data.

df.mean()       # Average
df.median()     # Middle value
df.mode()       # Most frequent
df.std()        # Standard deviation
df.var()        # Variance
df.min()        # Minimum
df.max()        # Maximum
df.count()      # Count
## Complete Statistical Summary
df.describe()

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Student': ['Amit', 'Ravi', 'Sita', 'Neha', 'Kiran'],
    'Math': [85, 90, 78, 92, 88],
    'Science': [88, 92, 80, 95, 85],
    'Gender': ['M', 'M', 'F', 'F', 'M']
}

df = pd.DataFrame(data)
```

```python
# Using Matplotlib
plt.plot(df['Student'], df['Math'])
plt.title("Math Scores")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
# bar chart using matplotlib
plt.bar(df['Student'], df['Science'])
plt.title("Science Scores")
plt.show()
```

```python
# using seaborn
sns.set(style="whitegrid")
sns.barplot(x='Student', y='Math', data=df)
plt.title("Math Scores - Seaborn")
plt.show()

```

```python
# seaborn scatter plot
sns.scatterplot(x='Math', y='Science', hue='Gender', data=df)
plt.title("Math vs Science")
plt.show()
```

```python
# seaborn boxplot
sns.boxplot(x='Gender', y='Math', data=df)
plt.title("Math Score Distribution by Gender")
plt.show()
```

```python
# Descriptive statistics
print("Descriptive Statistics:")
print(df.describe())
```
