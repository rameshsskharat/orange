```python
data_tuples = [
    ("Alice", 25, "New York"),
    ("Bob", 30, "Los Angeles"),
    ("Charlie", 35, "Chicago"),
    ("David", 40, "Houston"),
    ("Eve", 45, "Phoenix"),
    ("Frank", 50, "Philadelphia"),
    ("Grace", 55, "San Antonio"),
    ("Heidi", 60, "San Diego"),
    ("Ivan", 65, "Dallas"),
    ("Judy", 70, "San Jose"),
    ("Ken", 75, "Austin"),
    ("Leo",None, "Jacksonville"),
    ("Mallory", 85, "Fort Worth"),
    ("Nina", 90,None),
    (None, 95, "Charlotte")
]

df_tuples = pd.DataFrame(data=data_tuples,  columns=["Name","Age", "City"])
df_tuples
```

```python
# adding a new column
df_tuples["Age_After_5_Years"] = df_tuples["Age"] + 5
```

```python
# filtering rows based on a condition
df_filtered = df_tuples[df_tuples["Age"] > 30]
# calculating the mean age
mean_age = df_tuples["Age"].mean()
print("Mean Age:", mean_age)

# find max age record
max_age_record = df_tuples[df_tuples["Age"] == df_tuples["Age"].max()].values[0]
max_age_record

# grouping by city and calculating the average age
grouped = df_tuples.groupby("City")["Age"].mean()
nummber_of_persons_in_city = df_tuples.groupby("Name").count()
print(grouped)
print(nummber_of_persons_in_city)
# sorting by age
sorted_df = df_tuples.sort_values(by="Age", ascending=False)
print(sorted_df)
df_tuples.to_csv("people.csv", index=True)
```
