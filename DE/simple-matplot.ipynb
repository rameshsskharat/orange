```python
# !pip3 install matplotlib seaborn
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Example1
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales': [100, 150, 200, 180]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.plot(x='Month', y='Sales', kind='hist', color='blue')
plt.title("Monthly Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# example2
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [100, 150, 200, 180, 220],
    'Expenses': [80, 120, 160, 140, 170],
    'Profit': [20, 30, 40, 40, 50]
}

df = pd.DataFrame(data)
df.set_index('Month', inplace=True)
```

```python
# Line Plot (Trend Analysis)
df.plot(kind='line')
plt.title("Line Plot")
plt.show()
```

```python
# Bar Chart (Comparison)
df.plot(kind='bar')
plt.title("Bar Chart")
plt.show()
# Horizontal Bar Chart
df.plot(kind='barh')
plt.title("Horizontal Bar Chart")
plt.show()
```

```python
df.describe()
```

```python
# combine both charts in one plot
fig, ax1 = plt.subplots()
df['Sales'].plot(kind='bar', color='blue', ax=ax1, position=0, width=0.4)
df['Expenses'].plot(kind='bar', color='orange', ax=ax1, position=1, width=0.4)
ax1.set_title("Sales and Expenses Comparison")
ax1.set_xlabel("Month")
ax1.set_ylabel("Amount")
plt.show()
```

```python
# Histogram (Distribution)
df['Sales'].plot(kind='hist')
plt.title("Histogram of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Box Plot (Statistical Summary)
df.plot(kind='box')
plt.title("Box Plot")
plt.show()
```

```python
# Area Plot
df.plot(kind='area')
plt.title("Area Plot")
plt.show()
```

```python
# Scatter Plot (Relationship Between Two Variables)
df.plot(kind='scatter', x='Sales', y='Profit')
plt.title("Scatter Plot")
plt.show()
```

```python
# Pie Chart
df.loc['Jan'].plot(kind='pie', autopct='%1.1f%%')
plt.title("Pie Chart for January")
plt.ylabel('')
plt.show()
```
